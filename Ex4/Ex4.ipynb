```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Read dataset
df = pd.read_csv("StudentsPerformance.csv")

# Display dataset
print(df)

# Display first five records
print("\nFirst five records:")
display(df.head())

# Display shape
print("\nShape:")
print(df.shape)

# Display data types
print("\nData types:")
print(df.dtypes)
```

```python
# Statistical summary
print("\nStatistical summary:")
display(df.describe())

# Filter students with math marks greater than 80
print("\nStudents with math score > 80:")
display(df[df["math score"] > 80])

# Filter students who completed test preparation
print("\nStudents who completed test preparation:")
display(df[df["test preparation course"] == "completed"].head(20))

# Sort students by math marks
print("\nSorted by math score:")
display(df.sort_values("math score", ascending=False).head(20))

# Sort students by reading score
print("\nSorted by reading score:")
display(df.sort_values("reading score", ascending=False).head(20))

# Gender-wise average math score
print("\nGender-wise average math score:")
display(df.groupby("gender")["math score"].mean())

# Gender-wise average reading score
print("\nGender-wise average reading score:")
display(df.groupby("gender")["reading score"].mean())
```

```python
# Descriptive statistics of math scores
print("Mean Math Marks:", df["math score"].mean())
print("Median Math Marks:", df["math score"].median())
print("Minimum Math Marks:", df["math score"].min())
print("Maximum Math Marks:", df["math score"].max())
print("Standard Deviation:", df["math score"].std())

# Correlation matrix
print("\nCorrelation Matrix:")
display(df[["math score", "reading score", "writing score"]].corr())

print("Math-Reading Correlation:",
      df["math score"].corr(df["reading score"]))
```

```python
# Bar chart
plt.figure(figsize=(10,5))
plt.bar(
    ["Math", "Reading", "Writing"],
    df[["math score", "reading score", "writing score"]].mean()
)
plt.title("Average Student Scores")
plt.xlabel("Subject")
plt.ylabel("Average Score")
plt.show()

# Line chart
plt.figure(figsize=(10,5))
plt.plot(range(1, 21), df["math score"].head(20), marker="o")
plt.title("Math Scores of First 20 Students")
plt.xlabel("Student Record")
plt.ylabel("Math Score")
plt.grid()
plt.show()

# Scatter plot
plt.figure(figsize=(8,5))
plt.scatter(df["reading score"], df["math score"])
plt.title("Math Scores vs Reading Scores")
plt.xlabel("Reading Score")
plt.ylabel("Math Score")
plt.show()
```

```python
# Histogram
plt.figure(figsize=(8,5))
plt.hist(df["math score"], bins=10)
plt.title("Math Score Distribution")
plt.xlabel("Math Score")
plt.ylabel("Number of Students")
plt.show()

# Box plot
plt.figure(figsize=(8,5))
plt.boxplot(df["math score"])
plt.title("Math Score Box Plot")
plt.ylabel("Math Score")
plt.show()
```
