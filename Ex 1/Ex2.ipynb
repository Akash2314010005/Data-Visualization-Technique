```python
import os
import pandas as pd
import matplotlib.pyplot as plt

# Load dataset
file_path = "StudentsPerformance.csv"
df = pd.read_csv(file_path)

print("Dataset loaded successfully!")
```

```python
# First 5 records
print("\nFirst 5 rows:")
display(df.head())

# Last 5 records
print("\nLast 5 rows:")
display(df.tail())
```

```python
# Dataset information
print("\nDataset Information:")
df.info()
```

```python
# Descriptive statistics
print("\nDescriptive Statistics:")
display(df.describe())
```

```python
# Missing values
print("\nMissing Values:")
display(df.isnull().sum())
```

```python
# Gender distribution
print("\nGender Distribution:")
display(df["gender"].value_counts())

# Race/Ethnicity distribution
print("\nRace/Ethnicity Distribution:")
display(df["race/ethnicity"].value_counts())

# Parental Level of Education
print("\nParental Level of Education:")
display(df["parental level of education"].value_counts())

# Test Preparation Course
print("\nTest Preparation Course:")
display(df["test preparation course"].value_counts())
```

```python
# Correlation
print("\nCorrelation Matrix:")
display(df.corr(numeric_only=True))
```

```python
# Average scores
print("\nAverage Scores:")
display(df[["math score", "reading score", "writing score"]].mean())
```

```python
# Gender distribution plot
df["gender"].value_counts().plot(kind="bar", title="Gender Distribution")
plt.xlabel("Gender")
plt.ylabel("Number of Students")
plt.show()
```

```python
# Race/Ethnicity distribution plot
df["race/ethnicity"].value_counts().sort_index().plot(kind="bar", title="Race/Ethnicity Distribution")
plt.xlabel("Race/Ethnicity")
plt.ylabel("Number of Students")
plt.show()
```

```python
# Average score plot
means = df[["math score", "reading score", "writing score"]].mean()
means.plot(kind="bar", title="Average Student Scores")
plt.xlabel("Subject")
plt.ylabel("Average Score")
plt.show()
```

```python
# Correlation heatmap
score_cols = ["math score", "reading score", "writing score"]
corr = df[score_cols].corr()

plt.figure(figsize=(6, 5))
plt.imshow(corr, cmap="viridis", vmin=-1, vmax=1)
plt.xticks(range(len(score_cols)), ["Math", "Reading", "Writing"])
plt.yticks(range(len(score_cols)), ["Math", "Reading", "Writing"])
plt.title("Correlation Matrix")

for i in range(len(score_cols)):
    for j in range(len(score_cols)):
        plt.text(j, i, f"{corr.iloc[i, j]:.2f}", ha="center", va="center")

plt.colorbar()
plt.show()
```

```python
# Export files
df.to_csv("students_performance.csv", index=False)
df.to_excel("students_performance.xlsx", index=False)
df.to_json("students_performance.json", orient="records", indent=4)

print("CSV, Excel and JSON files created successfully.")
```
