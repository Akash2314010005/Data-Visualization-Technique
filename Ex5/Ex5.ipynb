```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("StudentsPerformance.csv")
first20 = df.head(20)
students = list(range(1, len(first20) + 1))
math_scores = first20["math score"]
reading_scores = first20["reading score"]
writing_scores = first20["writing score"]
avg_scores = [df["math score"].mean(), df["reading score"].mean(), df["writing score"].mean()]

print("Dataset shape:", df.shape)
display(df.head())
```

```python
# Program 1: Line Chart
plt.figure(figsize=(9,5))
plt.plot(students, math_scores, marker="o")
plt.title("Math Scores of First 20 Students")
plt.xlabel("Student Record")
plt.ylabel("Math Score")
plt.show()
```

```python
# Program 2: Bar Chart
plt.figure(figsize=(8,5))
plt.bar(["Math","Reading","Writing"], avg_scores)
plt.title("Average Student Scores")
plt.xlabel("Subject")
plt.ylabel("Average Score")
plt.show()
```

```python
# Program 3: Pie Chart
plt.figure(figsize=(7,7))
plt.pie(avg_scores, labels=["Math","Reading","Writing"], autopct="%1.1f%%")
plt.title("Subject Average Score Distribution")
plt.show()
```

```python
# Program 4: Histogram
plt.figure(figsize=(8,5))
plt.hist(df["math score"], bins=10)
plt.title("Math Score Distribution")
plt.xlabel("Math Score")
plt.ylabel("Frequency")
plt.show()
```

```python
# Program 5: Scatter Plot
plt.figure(figsize=(8,5))
plt.scatter(df["reading score"], df["math score"])
plt.title("Math Scores vs Reading Scores")
plt.xlabel("Reading Score")
plt.ylabel("Math Score")
plt.show()
```

```python
# Program 6: Box Plot
plt.figure(figsize=(8,5))
plt.boxplot(df["math score"])
plt.title("Math Score Box Plot")
plt.ylabel("Math Score")
plt.show()
```

```python
# Program 7: Multiple Line Charts
plt.figure(figsize=(10,5))
plt.plot(students, math_scores, label="Math", marker="o")
plt.plot(students, reading_scores, label="Reading", marker="o")
plt.plot(students, writing_scores, label="Writing", marker="o")
plt.title("Math, Reading and Writing Scores")
plt.xlabel("Student Record")
plt.ylabel("Score")
plt.legend()
plt.show()
```
