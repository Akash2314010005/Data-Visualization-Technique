```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Load Dataset
file_path = "StudentsPerformance.csv"
df = pd.read_csv(file_path)

# Select Important Columns
student_df = pd.DataFrame({
    "Gender": df["gender"].head(5).tolist(),
    "Race_Ethnicity": df["race/ethnicity"].head(5).tolist(),
    "Parental_Education": df["parental level of education"].head(5).tolist(),
    "Test_Preparation": df["test preparation course"].head(5).tolist(),
    "Math_Score": df["math score"].head(5).tolist(),
    "Reading_Score": df["reading score"].head(5).tolist(),
    "Writing_Score": df["writing score"].head(5).tolist()
})

print("Original DataFrame:")
display(student_df)
```

```python
# 1. First 5 Rows
print("\n1. First 5 rows:")
display(student_df.head())

# 2. Last 2 Rows
print("\n2. Last 2 rows:")
display(student_df.tail(2))

# 3. Shape
print("\n3. Shape:")
print(student_df.shape)

# 4. Column Names
print("\n4. Column Names:")
print(student_df.columns)

# 5. Data Types
print("\n5. Data Types:")
print(student_df.dtypes)

# 6. Statistical Summary
print("\n6. Statistical Summary:")
display(student_df.describe())
```

```python
# 7. Missing Values
print("\n7. Missing Values:")
display(student_df.isnull().sum())

# 8. Students who completed test preparation
print("\n8. Students who completed test preparation:")
completed = df[df["test preparation course"] == "completed"][[
    "gender",
    "race/ethnicity",
    "parental level of education",
    "test preparation course",
    "math score",
    "reading score",
    "writing score"
]].head(10)
display(completed)

# 9. Female Students
print("\n9. Female Students:")
female = df[df["gender"] == "female"][[
    "gender",
    "race/ethnicity",
    "parental level of education",
    "test preparation course",
    "math score",
    "reading score",
    "writing score"
]].head(10)
display(female)
```

```python
# 10. Visualization - Average Scores
avg_scores = df[["math score", "reading score", "writing score"]].mean()

plt.figure(figsize=(10, 5))
plt.bar(["Math", "Reading", "Writing"], avg_scores.values)
plt.title("Average Student Scores")
plt.xlabel("Subject")
plt.ylabel("Average Score")
plt.tight_layout()
plt.show()

# 11. Gender Type Distribution
gender_counts = df["gender"].value_counts()

plt.figure(figsize=(9, 5))
plt.bar(gender_counts.index, gender_counts.values)
plt.title("Gender Distribution")
plt.xlabel("Gender")
plt.ylabel("Number of Students")
plt.tight_layout()
plt.show()

# End
print("\nEDA inspection and filtering completed successfully.")
```
